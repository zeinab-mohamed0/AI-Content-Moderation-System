# Data Preprocessing

```python
def clean_text(text):
    text = text.lower()  # lowercase
    text = re.sub(r"http\S+", "", text)  # remove links
    text = re.sub(r"[^a-zA-Z\s]", "", text)  # remove punctuation & numbers
    text = re.sub(r"\s+", " ", text).strip()  # remove extra spaces
    return text
```

```python
df['cleaned_text'] = df['image descriptions'].apply(clean_text)
```

## Tokenization

```python
le = LabelEncoder()
df['label'] = le.fit_transform(df['Toxic Category'])
```

```python
df["cleaned_text"] = df["cleaned_text"].astype(str)
df["label"] = df["label"].astype(int)
```

```python
train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)
```

```python
train_dataset = Dataset.from_pandas(train_df)
test_dataset = Dataset.from_pandas(test_df)
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
def tokenize_function(batch):
    return tokenizer(
        batch["cleaned_text"],
        truncation=True,
        max_length=256
    )
```

```python
train_dataset = train_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)
```

```python
train_dataset = train_dataset.remove_columns(
    [col for col in train_dataset.column_names if col not in ["input_ids", "attention_mask", "label"]]
)

test_dataset = test_dataset.remove_columns(
    [col for col in test_dataset.column_names if col not in ["input_ids", "attention_mask", "label"]]
)
```

```python
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

# Model building

```python
base_model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=9
)
```

```python
lora_config = LoraConfig(
    task_type=TaskType.SEQ_CLS,
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=["q_lin", "v_lin"]
)

model = get_peft_model(base_model, lora_config)
model.print_trainable_parameters()
```

```python
accuracy = evaluate.load("accuracy")
f1 = evaluate.load("f1")
```

```python
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    acc = accuracy.compute(
        predictions=predictions,
        references=labels
    )["accuracy"]

    f1_score = f1.compute(
        predictions=predictions,
        references=labels,
        average="weighted"
    )["f1"]

    return {
        "accuracy": acc,
        "f1": f1_score
    }
```

```python
training_args = TrainingArguments(
    learning_rate=2e-4,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=5,
    weight_decay=0.01,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="steps",
    logging_steps=20,
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,
    report_to="none"
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)
```

# Evaluation

```python
trainer.train()
```

```python
results = trainer.evaluate()
print(results)
```

```python
model.save_pretrained("distilbert_model")

tokenizer.save_pretrained("distilbert_tokenizer")
```

# Pipeline

```python
def predict_text(text):
    # preprocess
    text = clean_text(text)

    # tokenize using DistilBERT tokenizer
    inputs = tokenizer(
        text,
        padding=True,
        max_length=256,
        return_tensors="pt"
    )

    # prediction
    outputs = model(**inputs)

    # logits
    logits = outputs.logits
    # predicted class index
    predicted_class = logits.argmax(dim=-1).item()

    # decode label
    label = le.inverse_transform([predicted_class])[0]

    return label
```
